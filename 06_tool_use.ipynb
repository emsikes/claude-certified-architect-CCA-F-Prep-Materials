```python
from anthropic import Anthropic
from dotenv import load_dotenv
import os
import pprint
```

```python
load_dotenv(override=True)

client = Anthropic()
model = "claude-sonnet-5"
```

A good tool should be narrow, well-named, and clearly describied.  
- Use required fields (schema), when they are not provided on input Claude will ask follow-up questions to get them
  - e.g. create_return_reguest needs all three fields:
    - order_id, item, reason
- Three ways to control tool use
  - By default (auto), Claude may either answer normally or request a tool 
    - {"type": "auto"}
  - We can specify that a tool must be used, but Clade decides which tool 
    - {"type": "any"}
  - We can pin a specific tool and enforce its usage 
    - {"type": "tool", "name": "lookup_order"}

A vague tool is much less reliable and Claude will have to guess its intent.



```python
tools = [
    {
        "name": "lookup_order",
        "description": "Look up an order by order ID and return the order details.",
        "input_schema": {
            "type": "object",
            "properties": {
                "order_id": {
                    "type": "string",
                    "description": "The customer's order ID."
                }
            },
            "required": ["order_id"]
        }
    }
]

message = client.messages.create(
    model=model,
    max_tokens=800,
    tools=tools,
    messages=[
        {
            "role": "user",
            "content": "Can I return order 12345?"
        }
    ]
)

pprint.pprint(message.stop_reason)
pprint.pprint(message.content)
```
