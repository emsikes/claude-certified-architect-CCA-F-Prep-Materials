```python
from anthropic import Anthropic
from dotenv import load_dotenv
import os
import pprint
```

```python
load_dotenv(override=True)

client = Anthropic()
model = "claude-sonnet-5"
```

Importand Techniques:
1. Clear and direct instructions
   * Tell Claude exactly what to do
   * Do not leave room for guessing the intent
2. Specific Guidelines:
   * Explain what good behavior looks like
   * Use explicit terms
3. Use XML style tags
   * Separate instructions, context, rules, and output format

Prompts Guide (soft behavior):
- Draft a polite response
- Classify the request
- Decide an order number is needed

Code enforces business rules (hard gurantees):
- Actual refund approval
- Permissions and access control
- Tools and application logic


```python
customer_message = "I want my money back."

prompt = f"""
<role>
You are a customer support assistant for ShopAssist AI
</role>
<task>
Write a short, helpful reply.
</task>
<context>
ShopAssist helps with orders, returns, refunds, and produt questions.
</context>
<rules>
- Be polite and calm.
- DO not approve a refund directly.
- Ask for the order number if it is missing
- Keep the response under 80 words.
</rules>
<customer_message>
{customer_message}
</customer_message>
<output_format>
Return only the final customer-facing response.
</output_format>
<success_criteria>
Short, calm, helpful - and asks for the order number.
</success_criteria>
"""

message = client.messages.create(
    model=model,
    max_tokens=800,
    output_config={"effort": "low"},
    messages=[{
        "role": "user",
        "content": prompt
    }]
)

pprint.pprint(message.content[0].text)
```
