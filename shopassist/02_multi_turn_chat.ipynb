```python
from anthropic import Anthropic
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)

client = Anthropic()
model = "claude-sonnet-5"
```

```python
message = client.messages.create(
    model=model,
    max_tokens=800,
    messages=[{"role": "user", "content": "I want to return my order."}],
)

print(message.stop_reason)
print("\n")
print(message.content[1].text)
```

```python
messages = [
    {
        "role": "user", 
        "content": "I want to return my order."
    },
    {
        "role": "assistant",
        "content": "Sure, I can help with that.  Can you share your order number?"
    },
    {
        "role": "user",
        "content": "My order number is 12345"
    }
]

message = client.messages.create(
    model=model,
    max_tokens=800,
    messages=messages
)

print(message.content[1].text)
```

```python
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=800,
        messages=messages
    )

    return message.content[1].text
```

```python
messages = []

add_user_message(messages, "I want to return my order.")

answer = chat(messages)

print("ShopAssist:", answer)

add_assistant_message(messages, answer)

add_user_message(messages, "My order number is 123456")

answer = chat(messages)

add_assistant_message(messages, answer)

messages
```

```python
system_prompt = """
You are ShopAssist AI, a helpful customer service support assistant for an online store.

Your job is to help customers with order questions, returns, refunds, shipping issues, and general inquiries.

Be concise, polite, and practical.

Do not promise a refund until the order is checked.

If you need more information, ask one clear question at a time, and wait for the user response before proceeding.
"""

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=800,
        system=system_prompt,
        messages=messages
    )

    return message.content[1].text


messages = []

add_user_message(
    messages,
    "Hi, I bought some headphones last week and the to not work.  I want my money back."
)

answer = chat(messages)

print("ShopAssist", answer)

add_assistant_message(messages, answer)

add_user_message(messages, "My order number is 123456")

answer = chat(messages)

print("ShopAssist", answer)

add_assistant_message(messages, answer)


messages
```
